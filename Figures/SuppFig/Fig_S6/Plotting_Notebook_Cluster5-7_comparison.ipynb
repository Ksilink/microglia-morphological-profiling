```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
alldata = pd.read_csv("Data/data_merged_cp_features_deep_labels.csv")

#  Plotting

# Set the context to 'talk'
sns.set_context('talk')
# Arial font
plt.rcParams["font.family"] = "Arial"

import seaborn as sns
import matplotlib.pyplot as plt

# Data to plot
x_col = "membership_Metadata_Deep"
y_col = "Intensity_IntegratedIntensity_OrigER"

# Determine category order (optional, ensures consistent ordering)
order = sorted(alldata[x_col].dropna().unique())

# Box plot (one box per category)
# Note: we don't set hue here to avoid duplicated/dodged boxes when hue==x
ax = sns.boxplot(
    data=alldata,
    x=x_col,
    y=y_col,
    hue=x_col,
    legend=False,
    order=order,
    palette="tab20",  # or any seaborn palette; can also pass a dict for custom colors
    showfliers=False  # hide outliers since we'll plot all points
)

# Overlay points
ax = sns.swarmplot(
    data=alldata,
    x=x_col,
    y=y_col,
    order=order,
    color="black",
    size=3,
    alpha=0.5
    )

plt.xlabel(x_col)
plt.ylabel(y_col)

ax.set_ylim([-1.5, 4])

# Show and save the figure
plt.tight_layout()

# Saving plot
plt.savefig(y_col + '.pdf', bbox_inches='tight')

plt.show()
```
